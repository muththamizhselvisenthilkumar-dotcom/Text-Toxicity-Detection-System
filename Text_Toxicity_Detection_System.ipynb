```python
import nltk
import re

from nltk.sentiment import SentimentIntensityAnalyzer

nltk.download('vader_lexicon')

sia = SentimentIntensityAnalyzer()
```

```python
text = """
This is a terrible and stupid comment. You are completely useless.
"""
```

```python
# Convert text to lowercase
lower_text = text.lower()

# Simple toxicity word list for demonstration
toxic_words = [
    "stupid",
    "idiot",
    "useless",
    "hate",
    "shut up",
    "fool",
    "loser",
    "dumb",
    "terrible",
    "worst"
]

# Find toxic words
detected_words = []

for word in toxic_words:
    if word in lower_text:
        detected_words.append(word)

# Calculate toxicity score
word_count = len(re.findall(r'\b\w+\b', text))

if word_count > 0:
    toxicity_score = (len(detected_words) / word_count) * 100
else:
    toxicity_score = 0

# Classify text
if len(detected_words) >= 2:
    result = "TOXIC ⚠️"
elif len(detected_words) == 1:
    result = "POTENTIALLY TOXIC ⚠️"
else:
    result = "NON-TOXIC ✅"

# Sentiment
sentiment_score = sia.polarity_scores(text)

if sentiment_score["compound"] >= 0.05:
    sentiment = "Positive 😊"
elif sentiment_score["compound"] <= -0.05:
    sentiment = "Negative 😞"
else:
    sentiment = "Neutral 😐"

# Display result
print("=" * 55)
print("          TEXT TOXICITY DETECTION SYSTEM")
print("=" * 55)

print("\nINPUT TEXT:")
print(text)

print("\n--- TOXICITY ANALYSIS ---")
print("Detected Words :", detected_words if detected_words else "None")
print("Toxicity Score :", round(toxicity_score, 2), "%")
print("Final Result   :", result)

print("\n--- SENTIMENT ANALYSIS ---")
print("Sentiment:", sentiment)

print("\n--- RECOMMENDATION ---")

if result == "TOXIC ⚠️":
    print("⚠️ This text may contain harmful or offensive language.")
    print("Recommendation: Review the message before posting.")
elif result == "POTENTIALLY TOXIC ⚠️":
    print("⚠️ This text contains potentially inappropriate language.")
    print("Recommendation: Consider using respectful language.")
else:
    print("✅ The text appears to be safe and non-toxic.")

print("\n" + "=" * 55)
```
